```python
import pandas as pd
import os
from glob import glob

folder = "/kaggle/input/datasets/rohanrao/nifty50-stock-market-data"

csv_files = glob(os.path.join(folder, "*.csv"))

csv_files = [
    f for f in csv_files
    if "stock_metadata" not in f.lower()
    and "nifty50_all" not in f.lower()
]

dfs = []

for file in csv_files:
    ticker = os.path.basename(file).replace(".csv", "")
    df = pd.read_csv(file)
    df["Symbol"] = ticker
    dfs.append(df)

individual_master = pd.concat(dfs, ignore_index=True)
```

```python
nifty_index = nifty50_v2.copy()

nifty_index.columns = [
    "Date",
    "NIFTY_Open",
    "NIFTY_High",
    "NIFTY_Low",
    "NIFTY_Close",
    "NIFTY_Volume",
    "NIFTY_Turnover"
]
```

```python
nifty_index["Date"] = pd.to_datetime(
    nifty_index["Date"]
)

nifty_index = nifty_index.sort_values(
    "Date"
).reset_index(drop=True)
```

```python
nifty_index["NIFTY_Return"] = (
    nifty_index["NIFTY_Close"]
    .pct_change()
)

nifty_index["NIFTY_MA20"] = (
    nifty_index["NIFTY_Close"]
    .rolling(20)
    .mean()
)

nifty_index["NIFTY_MA50"] = (
    nifty_index["NIFTY_Close"]
    .rolling(50)
    .mean()
)

nifty_index["NIFTY_Volatility_20"] = (
    nifty_index["NIFTY_Return"]
    .rolling(20)
    .std()
)
```

```python
print(nifty_fixed["Date"].dtype)
print(nifty_index["Date"].dtype)
```

```python
nifty_fixed["Date"] = pd.to_datetime(
    nifty_fixed["Date"]
)

nifty_index["Date"] = pd.to_datetime(
    nifty_index["Date"]
)
```

```python
print(nifty_fixed["Date"].dtype)
print(nifty_index["Date"].dtype)
```

```python
master_df = nifty_fixed.merge(
    nifty_index,
    on="Date",
    how="left"
)
```

```python
print(master_df.shape)

print(
    master_df[
        [
            "NIFTY_Close",
            "NIFTY_Return"
        ]
    ].isnull().sum()
)
```

```python
master_df.to_csv(
    "master_df.csv",
    index=False
)
```

```python
master_df['Turnover'] = master_df['Turnover'].fillna(
    master_df['Volume'] * master_df['Close']
)
```

```python
remaining = (
    master_df['Deliverable Volume'].isna()
    &
    master_df['%Deliverble'].isna()
)

remaining.sum()
```

```python
missing_stocks = (
    master_df.loc[
        master_df['Deliverable Volume'].isna(),
        'Symbol'
    ]
    .value_counts()
)

print(missing_stocks.head(20))
```

```python
master_df["Trades"] = master_df["Trades"].fillna(0)
```

```python
master_df['DV_Missing'] = (
    master_df['Deliverable Volume']
    .isna()
    .astype(int)
)
```

```python
master_df['%D_Missing'] = (
    master_df['%Deliverble']
    .isna()
    .astype(int)
)
```

```python
master_df['%Deliverble'] = (
    master_df.groupby('Symbol')['%Deliverble']
    .transform(
        lambda x: x.fillna(x.median())
    )
)
```

```python
master_df['Deliverable Volume'] = (
    master_df['Volume']
    * master_df['%Deliverble']
    / 100
)
```
